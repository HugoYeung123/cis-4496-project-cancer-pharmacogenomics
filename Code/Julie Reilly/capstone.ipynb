```python
cat_columns = ['TCGA_DESC','GDSC Tissue descriptor 1','GDSC Tissue descriptor 2','Cancer Type (matching TCGA label)','Microsatellite instability Status (MSI)',
                'Screen Medium','Growth Properties', 'CNA', 'Gene Expression', 'Methylation', 'TARGET']

f1[cat_columns] = f1[cat_columns].fillna("Unknown")
```

```python
#for f1
ncol = 3
#numerical features
num_col = ['LN_IC50', 'AUC', 'Z_SCORE']
num_features = min(8, len(numcol))
nrow = (num_features + ncol -1) // ncol

fig,axes = plt.subplots(nrow, ncol, figsize=(18, nrow*4))
axes = axes.flatten()

for i, col in enumerate(num_col[:num_features]):
    sns.histplot(data=f1, x=col, kde=True, bins=30, ax=axes[i])
    mean = f1[col].mean()
    median = f1[col].median()
    axes[i].axvline(mean, label = 'mean', color = 'green')
    axes[i].axvline(median, label = 'mean', color = 'pink')
```
